## MISIÓN 1: RETAIL – Evitar la fuga de clientes

La empresa ShopSmart quiere entender por qué los clientes abandonan.

Objetivo: Identificar variables importantes y reducir dimensionalidad.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

data_retail = pd.DataFrame({
    'visitas_web_mes': np.random.randint(1, 50, 100),
    'uso_descuento': np.random.randint(0, 2, 100),
    'satisfaccion': np.random.randint(1, 10, 100),
    'compras_previas': np.random.randint(1, 20, 100),
    'ingreso_mensual': np.random.randint(500, 5000, 100),
    'abandono_cliente': np.random.randint(0, 2, 100)
})

data_retail.head()
```

### Actividad: Correlación

Analiza qué variables están relacionadas con el abandono.

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr = data_retail.corr()

plt.figure()
sns.heatmap(corr, annot=True)
plt.title('Matriz de correlación')
plt.show()
```

**Preguntas:**
- ¿Qué variables se relacionan más con abandono_cliente?

  Las variables que muestran mayor relación con abandono_cliente suelen ser aquellas que reflejan la experiencia y comportamiento del usuario. La satisfacción del cliente destaca con una relación negativa importante, indicando que niveles bajos aumentan la probabilidad de abandono. También influyen las compras previas, ya que clientes más recurrentes tienden a quedarse. En menor medida, variables como el uso de descuentos o visitas web pueden aportar señales sobre intención o dependencia.

- ¿Eliminarías alguna variable?
  Diria que sería conveniente eliminar variables que no aporten información relevante al modelo, especialmente aquellas con correlaciones muy bajas respecto a la variable objetivo. Además, si existen variables altamente correlacionadas entre sí, se podría eliminar una de ellas para evitar redundancia. Esto ayuda a simplificar el modelo, mejorar su interpretabilidad y reducir ruido en el análisis sin perder capacidad predictiva.

### Actividad: Escalamiento

Aplica MinMax y Estandarización.

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd

features = ['visitas_web_mes','uso_descuento','satisfaccion','compras_previas','ingreso_mensual']

# Escalamiento MinMax
scaler_minmax = MinMaxScaler()
data_minmax = scaler_minmax.fit_transform(data_retail[features])

# Escalamiento Z-score
scaler_std = StandardScaler()
data_std = scaler_std.fit_transform(data_retail[features])

# Convertir a DataFrame para visualizar
df_minmax = pd.DataFrame(data_minmax, columns=features)
df_std = pd.DataFrame(data_std, columns=features)

print("Datos originales:")
display(data_retail[features].head())

print("\n Min-Max Scaling:")
display(df_minmax.head())

print("\n Z-score Scaling:")
display(df_std.head())
```

**Preguntas:**
- ¿Qué diferencias observas?
  La principal diferencia radica en cómo transforman los datos: MinMaxScaler los ajusta a un rango específico, generalmente entre 0 y 1, mientras que StandardScaler los centra en torno a una media de cero con desviación estándar de uno. El primero es más sensible a valores extremos, ya que estos afectan directamente los límites del rango. En cambio, la estandarización mantiene mejor la estructura de la distribución de los datos.

- ¿Cuál usarías para PCA?
  Para aplicar PCA es más adecuado usar StandardScaler, ya que este método depende de la varianza de los datos y requiere que todas las variables estén en una misma escala. Si no se estandarizan, aquellas con valores más grandes dominarían los componentes principales. Al estandarizar, se asegura que cada variable contribuya de forma equilibrada al análisis.

### Actividad: PCA

Reduce dimensiones y analiza la varianza.

```python
from sklearn.decomposition import PCA

pca = PCA() #Se crea un modelo de Análisis de Componentes Principales (PCA)
pca.fit(data_std) #Ajustar el modelo a los datos

explained_variance = pca.explained_variance_ratio_ #Obtener la varianza

plt.figure()
plt.plot(explained_variance, marker='o')
plt.title('Varianza explicada')
plt.show()
```

```python
import numpy as np

var_acumulada = np.cumsum(explained_variance)

plt.plot(var_acumulada, marker='o')
plt.title('Varianza acumulada')
plt.xlabel('Número de componentes')
plt.ylabel('Varianza acumulada')
plt.show()
```

**Preguntas:**
- ¿Cuántos componentes usarías?
  El número de componentes principales se elige con base en la varianza explicada acumulada, buscando conservar la mayor cantidad de información posible. En la práctica, se suele seleccionar el número de componentes que expliquen entre el 80% y el 95% de la varianza total. Generalmente, esto se logra con dos o tres componentes, lo que permite reducir significativamente la dimensionalidad sin perder información relevante.

- ¿Cómo lo explicarías a un gerente?
  PCA se puede explicar como una técnica que resume muchas variables en unas pocas más simples que capturan lo más importante de los datos. En lugar de analizar múltiples factores por separado, se crean indicadores clave que representan el comportamiento general del cliente. Esto facilita la toma de decisiones y permite trabajar con modelos más rápidos y fáciles de interpretar.

## MISIÓN 2: BANCA – Reducir riesgo financiero

El banco CreditSafe necesita mejorar sus decisiones de crédito.

**Objetivo**: Identificar variables clave.

```python
from sklearn.preprocessing import LabelEncoder

data_banco = pd.DataFrame({
    'historial_credito': np.random.choice(['bueno','malo'], 100),
    'tipo_empleo': np.random.choice(['fijo','temporal'], 100),
    'aprobado': np.random.randint(0,2,100)
})

le = LabelEncoder() #Codificador convierte texto → números

data_banco['historial_credito'] = le.fit_transform(data_banco['historial_credito']) #Transforma variables categóricas (bueno = 1, malo = 0)
data_banco['tipo_empleo'] = le.fit_transform(data_banco['tipo_empleo'])

data_banco.head()
```

### Actividad: Chi-cuadrado

Evalúa qué variable es más importante.

```python
from sklearn.feature_selection import chi2

X = data_banco[['historial_credito','tipo_empleo']]
y = data_banco['aprobado']

chi_scores = chi2(X, y)
chi_scores
```

### Actividad: Correlación de Pearson

Evalúa qué variable es más importante.

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr_banco = data_banco.corr()

plt.figure()
sns.heatmap(corr_banco, annot=True)
plt.title("Correlación de Pearson - Banca")
plt.show()
```

**Preguntas:**
- ¿Qué variable es más relevante?
  La variable más relevante seria aquella que presenta el valor más alto en la prueba, lo que indica mayor dependencia con la variable objetivo. En este contexto, el historial crediticio suele destacar porque refleja directamente el comportamiento pasado del cliente. Esto lo convierte en un factor clave para predecir la aprobación de un crédito.

- ¿Cómo ayuda esto al banco?
  En la correlación de Pearson, la variable más relevante es la que tiene el valor absoluto más alto en relación con la variable objetivo. Esto indica una relación más fuerte, ya sea positiva o negativa. Generalmente, variables como el historial crediticio o ingresos tienden a mostrar mayor influencia, mientras que otras pueden tener un impacto menor en la decisión final.
